```python
import pandas as pd

# 1. CARREGANDO OS DADOS
df = pd.read_csv("cancelamentos.csv")
```

```python
# 2. LIMPEZA DOS DADOS (TRATAMENTO)
df = df.drop(columns=["CustomerID"])
df = df.dropna()
df["cancelou"] = df["cancelou"].astype(int)

```

```python
# 3. ANÁLISE DOS CANCELAMENTOS (INSIGHTS)
total_clientes = len(df)
total_cancelados = df["cancelou"].sum()
total_ativos = total_clientes - total_cancelados
taxa_cancelamento = (total_cancelados / total_clientes) * 100

print("VISÃO GERAL DOS CLIENTES:")
print(f"Total de Clientes: {total_clientes}")
print(f"Clientes Ativos: {total_ativos}")
print(f"Clientes Cancelados: {total_cancelados}")
print(f"Taxa de Cancelamento Geral: {taxa_cancelamento:.2f}%\n")

# Analisando os gargalos da base
print("TAXA DE CANCELAMENTO POR CATEGORIA:")
print("\nPor Tipo de Contrato (%):")
print(df.groupby("duracao_contrato")["cancelou"].mean() * 100)

print("\nPor Ligações no Callcenter (%):")
print(df.groupby("ligacoes_callcenter")["cancelou"].mean() * 100)

print("\nPor Dias de Atraso (%):")
print(df.groupby("dias_atraso")["cancelou"].mean() * 100)
```

```python
# 4. CRIANDO NOVAS CATEGORIAS
status_texto = []
risco_cancelamento = []

for index, linha in df.iterrows():
    if linha["cancelou"] == 1:
        status_texto.append("Cancelado")
    else:
        status_texto.append("Ativo")
        
    if linha["ligacoes_callcenter"] > 4 or linha["dias_atraso"] > 20:
        risco_cancelamento.append("Alto Risco")
    else:
        risco_cancelamento.append("Baixo Risco")

df["status_cliente"] = status_texto
df["alerta_risco"] = risco_cancelamento
```

```python
# 5. EXPORTANDO PARA O POWER BI
df.to_csv("cancelamentos_pronto_bi.csv", index=False, encoding="utf-8-sig")

print("\n Processamento concluído!")
print("Arquivo 'cancelamentos_pronto_bi.csv' gerado com sucesso.")
```
